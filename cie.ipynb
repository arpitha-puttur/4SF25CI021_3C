```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("whitegrid")
```

```python
df = pd.read_csv("netflix_titles.csv")

df.head()
```

```python
# Display first five records
print("First 5 Records:")
display(df.head())
```

```python
# Display last five records
print("Last 5 Records:")
display(df.tail())
```

```python
# Number of rows and columns
print("Dataset Shape:", df.shape)
```

```python
# Column names
print("Column Names:")
print(df.columns)
```

```python
# Count null values in each column
print("Missing Values:")
print(df.isnull().sum())
```

```python
# Calculate null percentage
print("\nMissing Value Percentage:")
print((df.isnull().sum() / len(df)) * 100)
```

```python
plt.figure(figsize=(12, 6))

sns.heatmap(df.isnull(), cmap="viridis", cbar=False)

plt.title("Missing Values in Netflix Dataset")
plt.xlabel("Columns")
plt.ylabel("Records")

plt.show()
```

```python
# Remove duplicate records
df.drop_duplicates(inplace=True)

# Remove unnecessary columns
df.drop(
    columns=["show_id", "description"],
    inplace=True,
    errors="ignore"
)

# Remove unwanted spaces from text columns
for col in df.select_dtypes(include=["object", "str"]).columns:
    df[col] = df[col].str.strip()

display(df.head())
```

```python
df["date_added"] = pd.to_datetime(
    df["date_added"],
    errors="coerce"
)

# Convert release year into integer-compatible numeric data
df["release_year"] = pd.to_numeric(
    df["release_year"],
    errors="coerce"
)

print(df.dtypes)
```

```python
# Mean imputation for numerical columns
numerical_columns = df.select_dtypes(include=np.number).columns

for col in numerical_columns:
    if df[col].isnull().sum() > 0:
        df[col] = df[col].fillna(df[col].mean())

print("Missing values after mean imputation:")
print(df[numerical_columns].isnull().sum())
```

```python
# Mode imputation for categorical columns
categorical_columns = df.select_dtypes(
    include=["object", "str"]
).columns

for col in categorical_columns:
    if df[col].isnull().sum() > 0:
        df[col] = df[col].fillna(df[col].mode()[0])

print("Missing values after imputation:")
print(df.isnull().sum())
```

```python
# Extract year when content was added
df["year_added"] = df["date_added"].dt.year

# Extract month when content was added
df["month_added"] = df["date_added"].dt.month_name()

# Create content age column
df["content_age"] = 2026 - df["release_year"]

# Create duration category
df["duration_category"] = np.where(
    df["type"] == "Movie",
    "Movie Duration",
    "TV Show Seasons"
)

display(df.head())
```

```python
content_count = df["type"].value_counts()

print(content_count)

plt.figure(figsize=(7, 5))

content_count.plot(
    kind="bar",
    color=["tomato", "skyblue"]
)

plt.title("Number of Movies vs TV Shows")
plt.xlabel("Content Type")
plt.ylabel("Number of Titles")
plt.xticks(rotation=0)

plt.show()
```

```python
top_countries = (
    df["country"]
    .dropna()
    .str.split(", ")
    .explode()
    .str.strip()
    .value_counts()
    .head(10)
)

print(top_countries)

plt.figure(figsize=(12, 6))

top_countries.plot(
    kind="bar",
    color="orange"
)

plt.title("Top 10 Countries Producing Netflix Content")
plt.xlabel("Country")
plt.ylabel("Number of Titles")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
yearly_content = df["year_added"].value_counts().sort_index()

print(yearly_content)

plt.figure(figsize=(12, 6))

yearly_content.plot(
    kind="line",
    marker="o",
    color="green"
)

plt.title("Netflix Content Added Each Year")
plt.xlabel("Year")
plt.ylabel("Number of Titles")

plt.grid(True)
plt.show()
```

```python
genres = (
    df["listed_in"]
    .dropna()
    .str.split(", ")
    .explode()
    .str.strip()
    .value_counts()
    .head(10)
)

print(genres)

plt.figure(figsize=(12, 6))

genres.plot(
    kind="bar",
    color="purple"
)

plt.title("Top 10 Netflix Genres")
plt.xlabel("Genre")
plt.ylabel("Number of Titles")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
rating_count = df["rating"].value_counts().head(10)

plt.figure(figsize=(10, 6))

sns.barplot(
    x=rating_count.index,
    y=rating_count.values,
    color="cornflowerblue"
)

plt.title("Top 10 Netflix Content Ratings")
plt.xlabel("Rating")
plt.ylabel("Number of Titles")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python

movie_data = df[df["type"] == "Movie"]

plt.figure(figsize=(10, 6))

sns.histplot(
    movie_data["duration"].dropna(),
    bins=30,
    kde=True,
    color="coral"
)

plt.title("Distribution of Movie Durations")
plt.xlabel("Duration in Minutes")
plt.ylabel("Number of Movies")

plt.show()
```

```python
numeric_data = df.select_dtypes(include=np.number)

plt.figure(figsize=(8, 6))

sns.heatmap(
    numeric_data.corr(),
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap")
plt.show()
```

```python
print("========== NETFLIX DATA ANALYSIS ==========")

print("\nTotal Netflix Titles:", len(df))

print("\nMovies:", (df["type"] == "Movie").sum())

print("\nTV Shows:", (df["type"] == "TV Show").sum())

print("\nMost Common Genre:")
print(genres.index[0])

print("\nTop Content Producing Country:")
print(top_countries.index[0])

print("\nMost Common Rating:")
print(df["rating"].mode()[0])

print("\nAverage Release Year:")
print(round(df["release_year"].mean(), 2))

print("\n===========================================")
```
